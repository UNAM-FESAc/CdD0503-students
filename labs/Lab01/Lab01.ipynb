# Ejercicio 6. Dominancia diagonal

No todas las matrices garantizan la convergencia del método de Jacobi.

Una condición suficiente de convergencia es que la matriz sea **diagonalmente dominante por filas**.

Una matriz

$$
A=[a_{ij}]
$$

es diagonalmente dominante si para cada fila se cumple

$$
\|a_{ii}\|
\ge
\sum_{j\neq i}\|a_{ij}\|.
$$
