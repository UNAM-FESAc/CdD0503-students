## Synthesis and interpretation

- Convexity provides global guarantees that do not extend automatically to multimodal functions.
- Golden-section and Fibonacci searches rely on unimodality of the bracket.
- Polynomial interpolation creates local surrogate models; their minimizers need not minimize the original objective.
- Newton requires suitable curvature; gradient and conjugate-gradient methods exploit different information.
- Inverse-BFGS uses gradient differences to learn a variable metric while maintaining positive definiteness under the curvature condition.
