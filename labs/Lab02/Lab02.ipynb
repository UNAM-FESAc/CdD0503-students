## Funciones auxiliares

Las siguientes funciones implementan sustitución hacia adelante y hacia atrás. Se proporcionan como herramientas para concentrar el trabajo en las factorizaciones.


````python
def sustitucion_adelante(L, b):
    """Resuelve Ly=b para L triangular inferior."""
    L = np.asarray(L, dtype=float)
    b = np.asarray(b, dtype=float)
    n = len(b)
    y = np.zeros(n, dtype=float)

    for i in range(n):
        y[i] = (b[i] - L[i, :i] @ y[:i]) / L[i, i]

    return y


def sustitucion_atras(U, y):
    """Resuelve Ux=y para U triangular superior."""
    U = np.asarray(U, dtype=float)
    y = np.asarray(y, dtype=float)
    n = len(y)
    x = np.zeros(n, dtype=float)

    for i in range(n - 1, -1, -1):
        x[i] = (y[i] - U[i, i + 1:] @ x[i + 1:]) / U[i, i]

    return x
````
