Para Jacobi, todas las componentes nuevas deben calcularse usando únicamente el vector anterior $x^{(k)}$:

$$
x_i^{(k+1)}
=
\frac{1}{a_{ii}}
\left(
b_i-\sum_{j\neq i}a_{ij}x_j^{(k)}
\right).
$$

$$
x_i^{(k+1)}

\frac{1}{a_{ii}}
\left[
b_i
-\sum_{j=1}^{i-1}a_{ij}x_j^{(k+1)}
-\sum_{j=i+1}^{n}a_{ij}x_j^{(k)}
\right].
$$
